# HF Upload

```python
import os
from unsloth import FastLanguageModel
from huggingface_hub import login

HF_TOKEN = "HF_TOKEN"  # Replace with your actual Hugging Face token
login(HF_TOKEN)

print("Searching for the exact checkpoint path...")

checkpoint_path = None
for root, dirs, files in os.walk("/kaggle/input/"):
    if "checkpoint-4100" in root and "adapter_config.json" in files:
        checkpoint_path = root
        break

if checkpoint_path is None:
    print("Could not find the folder. Contents of /kaggle/input/:")
    print(os.popen("ls -R /kaggle/input/").read())
else:
    print(f"Found exact path: {checkpoint_path}")
    
    model, tokenizer = FastLanguageModel.from_pretrained(
        model_name = checkpoint_path, 
        max_seq_length = 2048,
        load_in_4bit = True,
    )

    print("Uploading the LoRA adapter to Hugging Face...")

    model.push_to_hub(
        "User_Name/Qwen2.5-Coder-7B-Codeforces-Tutor",
        token=HF_TOKEN
    )
    tokenizer.push_to_hub(
        "User_Name/Qwen2.5-Coder-7B-Codeforces-Tutor",
        token=HF_TOKEN
    )
    
    print("Adapter is safely uploaded to Hugging Face")
```
